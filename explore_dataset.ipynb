# AI Hub 71557 — 화자 중심 탐색 노트북 (v2)

**주 목적**: `speaker_id` 를 선택하면 해당 화자의 메타정보와 무작위 오디오 샘플을 함께 확인.

**필요 입력**: `metadata.csv` (= `build_metadata.py` 실행 결과).

**서버 이전 안전**: `audio` 컬럼은 `base_dir` 기준 상대 경로. 노트북 상단에서 `OVERRIDE_BASE_DIR` 만 바꾸면 경로 전체 갱신.

```python
import pandas as pd
from pathlib import Path
from IPython.display import Audio, display, HTML
import ipywidgets as W

pd.set_option('display.max_colwidth', 120)
pd.set_option('display.width', 200)

META_CSV = './meta/metadata.csv'
# 서버 이전 시: 이 변수에 새 base_dir 를 넣으면 audio_path 가 자동으로 갱신됨.
# None 이면 CSV 의 base_dir/audio_path 를 그대로 사용.
OVERRIDE_BASE_DIR = None  # 예: '/data2/aihub/'

df = pd.read_csv(META_CSV)

if OVERRIDE_BASE_DIR is not None:
    new_base = OVERRIDE_BASE_DIR.rstrip('/') + '/'
    df['base_dir']   = new_base
    df['audio_path'] = new_base + df['audio'].astype(str)
    print(f'[OVERRIDE] base_dir → {new_base}')

print(f'총 {len(df):,}개 발화 로드')
print(f'  Training  : {(df.split == "Training").sum():,}')
print(f'  Validation: {(df.split == "Validation").sum():,}')
print(f'  화자 수    : {df.speaker_id.nunique()}')
df.head(3)
```

## 1. 화자 인덱스 (개요표)

전체 화자를 한눈에 — `speaker_id`, 성별/연령/직업, 발화 수, 총 시간, 평균 길이.

```python
def speaker_index(df):
    g = (df.groupby('speaker_id', dropna=True)
         .agg(sex=('sex','first'), age=('age','first'), job=('job','first'),
              n=('audio_duration','count'),
              total_sec=('audio_duration','sum'),
              mean_sec=('audio_duration','mean')))
    g['total'] = g.total_sec.map(lambda s: f'{int(s)//3600:d}:{(int(s)%3600)//60:02d}:{int(s)%60:02d}')
    g['mean_sec'] = g.mean_sec.round(2)
    g['n'] = g.n.astype(int)
    return g[['sex','age','job','n','total','mean_sec']].sort_values('n', ascending=False)

idx = speaker_index(df)
print(f'화자 {len(idx)} 명')
idx.head(20)
```

## 2. 핵심 함수: `inspect_speaker(spk_id)` 와 `play(rows)`

- `inspect_speaker('SPK014')` → 그 화자의 메타정보 카드 + N개 샘플 (필요시 필터)
- `play(rows)` → DataFrame 의 각 행을 카드+오디오 위젯으로 표시

```python
def _fmt_hms(sec):
    sec = int(sec); h, r = divmod(sec, 3600); m, s = divmod(r, 60)
    return f'{h:d}:{m:02d}:{s:02d}'

def play(rows, max_n=5):
    '''행들의 메타정보 + 오디오 위젯을 표시.'''
    if isinstance(rows, pd.Series):
        rows = rows.to_frame().T
    if len(rows) == 0:
        print('(매칭되는 샘플 없음)')
        return
    for i, row in rows.head(max_n).reset_index(drop=True).iterrows():
        html = f'''
        <div style="border-left:3px solid #4a90e2; padding:8px 14px; margin:10px 0; background:#fafbfc;">
          <div style="color:#666; font-size:0.88em;">
            <b>[{i+1}] {row['filename']}</b> &nbsp;|&nbsp; {row['speaker_id']} 
            ({row['age']}/{row['sex']}/{row['job']}) 
            &nbsp;|&nbsp; <b>{row['press']}</b>/{row['press_field']}/{row['press_date']} 
            &nbsp;|&nbsp; <code>{row['sentence_type']}</code> 
            &nbsp;|&nbsp; {row['audio_duration']:.2f}s
          </div>
          <div style="margin-top:6px; font-size:1.05em; line-height:1.5;">📝 {row['text']}</div>
          <div style="margin-top:4px; color:#888; font-size:0.82em;">
            🏷️ {row['keyword']}
          </div>
        </div>
        '''
        display(HTML(html))
        wav = row['audio_path']
        if isinstance(wav, str) and Path(wav).is_file():
            display(Audio(wav))
        else:
            print(f'   [audio 없음: {wav}]')


def inspect_speaker(
    speaker_id,
    n=5, random_state=None,
    sentence_type=None,  # '작문형'|'요약형'|'완전직접인용형'
    press_field=None,    # '정치'|'경제'|... 
    press=None,          # 'KBS'|'SBS'|...
    text=None,           # 부분 일치
    duration_min=None, duration_max=None,
    show_card=True,
):
    '''speaker_id 의 메타정보 카드를 표시하고 (선택) 필터 적용 후 N개 샘플 재생.'''
    sub = df[df['speaker_id'] == speaker_id]
    if len(sub) == 0:
        print(f'[ERROR] speaker_id "{speaker_id}" 가 metadata 에 없습니다.')
        return None

    if show_card:
        meta = sub.iloc[0]
        total = sub['audio_duration'].sum()
        # 분야·언론사·문장유형 top 3 요약
        def _top(col, k=3):
            vc = sub[col].value_counts().head(k)
            return ', '.join(f'{i}({c})' for i, c in vc.items())
        display(HTML(f'''
        <div style="background:linear-gradient(90deg,#e8f0fe,#fff); padding:14px 18px;
                    border-radius:8px; margin:10px 0; border-left:4px solid #1a73e8;">
          <h3 style="margin:0 0 8px 0;">🎙️ {speaker_id} 
            <span style="font-weight:normal; color:#666; font-size:0.7em;">
              ({meta['age']} / {meta['sex']} / {meta['job']})
            </span>
          </h3>
          <table style="font-size:0.92em; border-collapse:collapse;">
            <tr><td style="padding:2px 12px 2px 0; color:#666;">전체 발화</td>
                <td><b>{len(sub):,}</b> 개</td></tr>
            <tr><td style="padding:2px 12px 2px 0; color:#666;">총 시간</td>
                <td><b>{_fmt_hms(total)}</b> &nbsp;({total:,.0f} 초)</td></tr>
            <tr><td style="padding:2px 12px 2px 0; color:#666;">평균 길이</td>
                <td>{sub.audio_duration.mean():.2f} 초 
                    (min {sub.audio_duration.min():.2f} / max {sub.audio_duration.max():.2f})</td></tr>
            <tr><td style="padding:2px 12px 2px 0; color:#666;">언론사 (top3)</td>
                <td>{_top('press')}</td></tr>
            <tr><td style="padding:2px 12px 2px 0; color:#666;">분야 (top3)</td>
                <td>{_top('press_field')}</td></tr>
            <tr><td style="padding:2px 12px 2px 0; color:#666;">문장유형</td>
                <td>{_top('sentence_type', 3)}</td></tr>
          </table>
        </div>
        '''))

    # 필터
    if sentence_type is not None: sub = sub[sub['sentence_type'] == sentence_type]
    if press_field   is not None: sub = sub[sub['press_field']   == press_field]
    if press         is not None: sub = sub[sub['press']         == press]
    if text          is not None: sub = sub[sub['text'].fillna('').str.contains(text, regex=False)]
    if duration_min  is not None: sub = sub[sub['audio_duration'] >= duration_min]
    if duration_max  is not None: sub = sub[sub['audio_duration'] <= duration_max]

    if len(sub) == 0:
        print('(필터 조건에 맞는 샘플이 없습니다.)')
        return None

    sample = sub.sample(n=min(n, len(sub)), random_state=random_state)
    play(sample, max_n=n)
    return sample.reset_index(drop=True)

print('inspect_speaker() / play() 정의 완료')
```

## 5. 인터랙티브 위젯: 드롭다운으로 화자 골라보기

```python
speakers = sorted(df['speaker_id'].dropna().unique())

ui_spk    = W.Dropdown(options=speakers, description='화자')
ui_n      = W.IntSlider(value=3, min=1, max=15, description='샘플 수')
ui_seed   = W.IntText(value=0, description='seed')
ui_stype  = W.Dropdown(
    options=[('(전체)', None)] + [(s, s) for s in sorted(df['sentence_type'].dropna().unique())],
    description='문장유형')
ui_field  = W.Dropdown(
    options=[('(전체)', None)] + [(s, s) for s in sorted(df['press_field'].dropna().unique())],
    description='분야')
ui_press  = W.Dropdown(
    options=[('(전체)', None)] + [(s, s) for s in sorted(df['press'].dropna().unique())],
    description='언론사')
ui_text   = W.Text(value='', description='텍스트', placeholder='부분 일치 검색')
ui_dmin   = W.FloatText(value=0,   description='min(초)')
ui_dmax   = W.FloatText(value=40,  description='max(초)')
btn       = W.Button(description='검색 & 재생', button_style='primary',
                     icon='play')
out       = W.Output()

def _on_click(_):
    out.clear_output()
    with out:
        inspect_speaker(
            ui_spk.value,
            n=ui_n.value, random_state=ui_seed.value,
            sentence_type=ui_stype.value,
            press_field=ui_field.value,
            press=ui_press.value,
            text=(ui_text.value or None),
            duration_min=(ui_dmin.value if ui_dmin.value > 0 else None),
            duration_max=(ui_dmax.value if ui_dmax.value < 999 else None),
        )

btn.on_click(_on_click)

display(W.VBox([
    W.HBox([ui_spk, ui_n, ui_seed]),
    W.HBox([ui_stype, ui_field, ui_press]),
    W.HBox([ui_text]),
    W.HBox([ui_dmin, ui_dmax, btn]),
    out,
]))
```
